```python

import random
import math

def calculate_conflicts(state):
    conflicts = 0
    n = len(state)

    for i in range(n):
        for j in range(i + 1, n):
            if state[i] == state[j]:
                conflicts += 1
            elif abs(state[i] - state[j]) == abs(i - j):
                conflicts += 1

    return conflicts


def simulated_annealing_n_queens(n, initial_temp=100.0, cooling_rate=0.995, max_iterations=100000):

    current_state = [random.randint(0, n - 1) for _ in range(n)]
    current_conflicts = calculate_conflicts(current_state)
    temp = initial_temp

    for iteration in range(max_iterations):

        if current_conflicts == 0:
            return current_state, iteration

        if temp < 0.00001:
            break

        random_col = random.randint(0, n - 1)
        original_row = current_state[random_col]

        available_rows = [r for r in range(n) if r != original_row]

        if not available_rows:
            continue

        random_row = random.choice(available_rows)

        neighbor_state = list(current_state)
        neighbor_state[random_col] = random_row

        neighbor_conflicts = calculate_conflicts(neighbor_state)

        delta_e = neighbor_conflicts - current_conflicts

        if delta_e < 0:
            current_state = neighbor_state
            current_conflicts = neighbor_conflicts

        else:
            acceptance_probability = math.exp(-delta_e / temp)

            if random.random() < acceptance_probability:
                current_state = neighbor_state
                current_conflicts = neighbor_conflicts

        temp *= cooling_rate

    if current_conflicts == 0:
        return current_state, max_iterations

    return None, max_iterations


def print_board(state):

    if not state:
        print("No solution found.")
        return

    n = len(state)

    for row in range(n):
        for col in range(n):
            if state[col] == row:
                print(" Q ", end="")
            else:
                print(" . ", end="")
        print()


if __name__ == "__main__":

    N = 8

    print(f"Solving {N}-Queens using Simulated Annealing...")

    solution, iterations = simulated_annealing_n_queens(N)

    if solution:
        print("\nSolution found within schedule limitations:\n")
        print_board(solution)

        print("\nState Array Representation:", solution)
        print("Iterations:", iterations)

    else:
        print("\nFailed to find a solution within the cooling schedule limit.")


print("Anitej Padyana 1wn24cs042")
```
